```python
# EJERCICIO 5: CLASIFICACION DE TRIANGULOS

import math

a = float(input("Ingrese el lado A: "))
b = float(input("Ingrese el lado B: "))
c = float(input("Ingrese el lado C: "))

if a + b <= c or a + c <= b or b + c <= a:
    print("Los lados no forman un triangulo.")

elif a == b and b == c:
    perimetro = a + b + c
    print("El triangulo es EQUILATERO.")
    print("Su perimetro es:", perimetro)

elif a == b:
    print("El triangulo es ISOSCELES.")
    print("Los lados A y B son iguales.")

elif a == c:
    print("El triangulo es ISOSCELES.")
    print("Los lados A y C son iguales.")

elif b == c:
    print("El triangulo es ISOSCELES.")
    print("Los lados B y C son iguales.")

else:
    s = (a + b + c) / 2
    area = math.sqrt(s * (s-a) * (s-b) * (s-c))
    print("El triangulo es ESCALENO.")
    print("Su area es:", area)
```

```python
# Ejercicio 3: Clasificación de números respecto a X

N = int(input("Ingrese la cantidad de números: "))
numeros = []

for i in range(N):
    numero = float(input("Ingrese un número: "))
    numeros.append(numero)

X = float(input("Ingrese el valor de X: "))

menores = 0
mayores = 0
iguales = 0
entre = 0
ninguna = 0

for numero in numeros:
    if numero < X:
        menores += 1
    elif numero > X:
        mayores += 1
    else:
        iguales += 1

    if 1 <= numero <= X:
        entre += 1

    if not (numero < X or numero > X or numero == X or 1 <= numero <= X):
        ninguna += 1

print("Porcentaje de menores que X:", menores / N * 100, "%")
print("Porcentaje de mayores que X:", mayores / N * 100, "%")
print("Porcentaje iguales a X:", iguales / N * 100, "%")
print("Porcentaje entre 1 y X:", entre / N * 100, "%")
print("Porcentaje que no cumple ninguna condición:", ninguna / N * 100, "%")
```
